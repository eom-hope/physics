```python
import math #수학 함수를 사용하기 위해서 라이브러리 추가
```

호도법

```python
angle_degree = 30

angle_radian = angle_degree * math.pi / 180

print("angle_degree: ", angle_radian)
```

```python
math.cos(angle_radian)
```

```python
math.sin(angle_radian)
```

```python
math.tan(angle_radian)
```

직교좌표를 극좌표로 변환

```python
x, y = 1.0, 2.0
```

```python
r = math.sqrt(x**2 + y**2)
theta_raidan = math.atan2(y, x)

print(r, theta_raidan)
```

극좌표를 직교 좌표로 변환

```python
r=10
angle_degree=30
```

```python
y=r+math.cos(math.radians(angle_degree))
x=angle_degree+math.tan(math.radians(angle_degree))

print(x,y)
```
